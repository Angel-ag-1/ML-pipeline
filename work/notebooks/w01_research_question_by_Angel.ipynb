```python
# 3. Quick look at the data (2-3 real numbers)
#Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.

import pandas as pd

df = pd.read_csv("/content/ML-pipeline/data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nTrend Direction:")
print(df["trend_direction"].value_counts())

print("\nAverage CTR:", round(df["ctr"].mean(), 4))

print("Average Engagement Rate:", round(df["engagement_rate"].mean(), 4))

# The starter dataset contains 30,000 webpages with 44 columns of old SEO information. The dataset includes different performance trends with 16,262 pages showing a downward trend. This provides enough examples to explore whether old SEO data can help prioritise pages for content review. The average CTR is 0.5107 and the average engagement rate is 2.5345. This is showing that the dataset contains differences in webpage performance that can be analysed.
```
